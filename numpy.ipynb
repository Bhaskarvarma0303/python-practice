<h2 >Ways to create arrays</h2>

<h4>1.arange

np.arange(start,stop,step)  #like for loop

```python
a=np.arange(1,10,2)
```

```python
a
```

```python
plt.plot(a)
```

<h4>2.linspace</h4>

np.linspace(start,stop,no.of items)  #automatically divides based of no.of items

```python
a=np.linspace(0,10,5)
```

```python
a
```

```python
plt.plot(a)
```

<h4>3.logspace

same as linspace but generates exponents

```python
a=np.logspace(0,5,10,base=3,dtype=int)
```

```python
a
```

```python
plt.plot(a)
```

<h4>4.Identity array

```python
a=np.identity(3)
```

```python
a
```

```python
plt.plot(a)
```

here three lines represents three rows of matrix
